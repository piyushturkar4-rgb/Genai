```python
#Data Loading first step
from pathlib import Path
from langchain_community.document_loaders import TextLoader
from langchain_core.documents import Document

base_dir = Path.cwd()
candidates = [base_dir / "data.txt", Path("e:/Gen AI/data.txt"), Path("data.txt")]
text_path = next((p for p in candidates if p.exists()), base_dir / "data.txt")

print("resolved_path:", text_path)
print("exists:", text_path.exists())
print("size:", text_path.stat().st_size if text_path.exists() else None)
print("raw_preview:", repr(text_path.read_text(encoding="utf-8")[:200]) if text_path.exists() else None)

loader = TextLoader(str(text_path), encoding="utf-8")
text_document = loader.load()

print("loader_docs:", len(text_document))
print("loader_preview:", repr(getattr(text_document[0], "page_content", "")[:200]) if text_document else None)

if not text_document or not any(getattr(doc, "page_content", "").strip() for doc in text_document):
    text_content = text_path.read_text(encoding="utf-8")
    text_document = [Document(page_content=text_content, metadata={"source": str(text_path)})]

text_document
```

```python
# Keep the loaded documents in a variable for later cells
text_document
```

```python
#Text splitter
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document

textSplitter = RecursiveCharacterTextSplitter(chunk_size=200, chunk_overlap=20)
finalDocuments = textSplitter.split_documents(text_document)

# Fallback if the splitter produces no chunks
if not finalDocuments and text_document:
    combined_text = "\n".join(
        doc.page_content for doc in text_document if getattr(doc, "page_content", "")
    )
    finalDocuments = [Document(page_content=combined_text, metadata={"source": "data.txt"})]

finalDocuments
```

```python
texts =[doc.page_content for doc in finalDocuments]
```

```python
##Huggingface embedding
import os 
from dotenv import load_dotenv
load_dotenv()
```

```python
os.environ['HF_Token']=os.getenv('HF_Token')
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
print(embeddings)
```

```python
from langchain_community.vectorstores import FAISS

if not finalDocuments:
    raise ValueError("No document chunks were generated. Check the source text file or splitter settings.")

db = FAISS.from_documents(finalDocuments, embeddings)
db
```

```python
query = input("Enter your question : ")
docs = db.similarity_search(query)
for i in range (len(docs)):
    print(docs[i].page_content)
```
